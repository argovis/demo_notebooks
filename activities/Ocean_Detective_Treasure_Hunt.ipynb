### **Learning goals**

By completing this activity, you will be able to:

* Interpret vertical temperature and salinity profiles and infer their effects on seawater density.
* Explain how atmospheric circulation, river discharge, and sea ice processes influence ocean salinity.
* Explain how temperature and salinity contribute to ocean density stratification and vertical mixing.
* Explain why relatively dense seawater can sink beneath less dense seawater.

### **Instructions**

At each stop, explore Argo temperature and salinity profiles in the specified region.

* **Collect clues:** For each question, choose an answer and record its corresponding **letter**, in order. The letters will reveal secret keywords.
* **Think and discuss:** Write down a brief explanation of your observations. These questions do not produce letters, but you will use your answers during the class discussion.
* **Unlock treasures:** Submit the keywords you discover through Canvas.

**Remember:** Colder and saltier seawater is generally denser than warmer and fresher seawater. Stable density stratification occurs when less dense water overlies denser water.

This notebook uses the Argovis web app and database. Argovis provides an API that indexes and distributes numerous oceanographic datasets with detailed query parameters, enabling you to search and download only and exactly the data of interest.

## Setup: how to use the Argovis website

No coding is needed for this activity: everything happens in your web browser at [argovis.colorado.edu/argo](https://argovis.colorado.edu/argo). Stops 2–4 also have a shortcut link that opens the right region directly, in case you are short on time.

**API key (optional).** In order to allocate Argovis's limited computing resources fairly, users are encouraged to register and request a free API key at [https://argovis-keygen.colorado.edu/](https://argovis-keygen.colorado.edu/). Treat your API key like a password. The website also works with the default `guest` key, but may be slower when many people use it at once.

**Step 1: find profiles on the map.** Set the start and end dates, then click the polygon button on the map and draw a shape around your region. Each dot is one Argo profile. Click a dot to see its ID, location and date, plus links to its plots.



**Step 2: open a plot.** **Profile Page** shows the profile you clicked. **Regional Selection Page** shows all the profiles in your polygon.

**Step 3: set the axes.** Under **Axis Controls**, choose `temperature` or `salinity` for the x-axis variable and keep `pressure` on the y-axis. Pressure in decibars is approximately depth in meters. To zoom in, type the limits in the **min** and **max** boxes and **press Return**: the plot does not update until you do. For example, use y-axis max = 200 for the upper 200 m.
